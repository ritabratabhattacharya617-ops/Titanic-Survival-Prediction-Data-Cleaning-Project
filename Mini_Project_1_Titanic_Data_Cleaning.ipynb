# Mini Project 1 — Titanic Survival Prediction: Data Cleaning Project

**Week 1: ML Fundamentals + Data Preprocessing**

This project completes the exact mini-project tasks listed in the supplied course PDF: clean missing data, encode `Sex` and `Embarked`, visualize age distribution, and export the cleaned dataset as a new CSV.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import LabelEncoder, OneHotEncoder

df = pd.read_csv("train.csv")
print("Shape:", df.shape)
df.head()
```

```python
print("Missing values before cleaning:")
print(df.isna().sum())

df["Age"] = df["Age"].fillna(df["Age"].median())
df["Fare"] = df["Fare"].fillna(df["Fare"].median())
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

print("\nMissing values after cleaning:")
print(df.isna().sum())
```

```python
# Encode Sex using LabelEncoder
label_encoder = LabelEncoder()
df["Sex_encoded"] = label_encoder.fit_transform(df["Sex"])

print("Sex mapping:")
print(dict(zip(label_encoder.classes_, label_encoder.transform(label_encoder.classes_))))
```

```python
# Encode Embarked using OneHotEncoder
one_hot = OneHotEncoder(handle_unknown="ignore", sparse_output=False)
embarked_encoded = one_hot.fit_transform(df[["Embarked"]])
embarked_df = pd.DataFrame(
    embarked_encoded,
    columns=one_hot.get_feature_names_out(["Embarked"]),
    index=df.index
)

df_clean = pd.concat([df, embarked_df], axis=1)
df_clean.head()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["Age"], bins=30)
plt.title("Titanic Passenger Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.tight_layout()
plt.show()
```

```python
output_path = "titanic_cleaned.csv"
df_clean.to_csv(output_path, index=False)
print("Saved:", output_path)
print("Final shape:", df_clean.shape)
```

## Result

The cleaned CSV contains the original Titanic fields plus `Sex_encoded` and one-hot encoded `Embarked` columns. Missing `Age`, `Fare`, and `Embarked` values were handled before encoding.
